```python
import numpy as np
import nibabel as nib
import SimpleITK as sitk
import os
import glob
from sklearn.metrics import mean_squared_error as MSE
import pandas as pd
import json
```

```python

test_folder="/home/changh/Documents/2023_DosePredcition/validation/91_f0_e575_test"
save_folder="/home/changh/Documents/2023_DosePredcition/validation/91_f0_e575_test/csv"
tests=sorted(glob.glob(os.path.join(test_folder,"*.nii.gz")))
```

```python
if not os.path.isdir(save_folder):
    os.makedirs(save_folder)
```

```python

for test in tests:
    file=test.split("/")[-1]
    patient_num=file.split(".")[0][1:4]
    # patient_path=os.path.join(save_folder,"pt_"+str(patient_num))
    # if not os.path.isdir(patient_path):
    #     os.makedirs(patient_path)

    pred=sitk.ReadImage(os.path.join(test_folder,file))
    pred_np=sitk.GetArrayFromImage(pred).flatten()
    #col1=np.arange(0,len(pred_np))
    col2=pred_np
    dict={'data':col2}
    df = pd.DataFrame(dict)
    df.to_csv(os.path.join(save_folder,"pt_"+str(patient_num)+".csv"))


    
    

    
```
