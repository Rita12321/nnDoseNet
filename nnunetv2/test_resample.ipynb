```python
import pydicom
import numpy as np
import SimpleITK as sitk
import nibabel as nib
import os
import threading
import glob
import shutil
import math as m
```

```python
def resample( image: sitk.Image):
    ref=sitk.ReadImage("/home/changh/Documents/2023_DosePredcition/Data/open-kbp-master/nifti_data_test/labelsTr/0001.nii.gz")

    resample = sitk.ResampleImageFilter()
    image.SetSpacing(ref.GetSpacing())
    image.SetDirection(ref.GetDirection())
    resample.SetReferenceImage(ref)
    resample.SetInterpolator(sitk.sitkNearestNeighbor)
    
    return resample.Execute(image)

# def resample(image: sitk.Image,transform=sitk.Euler3DTransform(),
#                  interpolator = sitk.sitkNearestNeighbor,default_value=0):

#         reference_image=sitk.ReadImage("/home/changh/Documents/2023_DosePredcition/Data/open-kbp-master/nifti_data_test/labelsTr/0001.nii.gz")
    
#         return sitk.Resample(image, reference_image, transform,
#                                 interpolator, default_value) 
```
